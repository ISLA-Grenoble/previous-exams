# Part 1: Multiple Choice Questions (8 points)

## Question 1 — Robustness to outliers (1 point)

We have a classification problem on linearly separable data with an outlier (a point very far from the other data points). We train LDA, logistic regression, and 1-NN on this dataset and test on a clean test set. We then remove the outlier and retrain.

**After retraining, which classifier will NOT change its decision boundary around the test points?**

- (A) Logistic regression.
- **(B) 1-nearest-neighbors classifier.** ✓
- (C) LDA.
- (D) None of them.

**Answer: B**

**Reason:** In 1-NN, the decision boundary is determined solely by the positions of the training points. If the outlier is very far from the test points, it does not act as the nearest neighbour to any test point — removing it does not change the boundaries around those test points. LDA and logistic regression both use global statistics (mean, covariance, likelihood) that are affected by any single point, so their boundaries shift when the outlier is removed.

## Question 2 — Bias-variance decomposition (1 point)

Data: $y = x + \varepsilon$, $x \sim \text{Uniform}[0,1]$, $\varepsilon \sim \mathcal{N}(0,1)$.
- Model A: trained quadratic $g_A(x) = \beta_0 + \beta_1 x + \beta_2 x^2$
- Model B: constant $g_B(x) = 1/2$

**Compared to model B, model A has:**

- (A) Higher bias, higher variance.
- **(B) Lower bias, higher variance.** ✓
- (C) Higher bias, lower variance.
- (D) Lower bias, lower variance.

**Answer: B**

**Reason:** Model B is a fixed constant and cannot fit $y = x$ well → high bias. Model A is flexible (quadratic) and can approximate the true linear relationship → lower bias. However, a quadratic model with learned parameters varies more across different training datasets → higher variance. This is the classic bias-variance trade-off.

## Question 3 — Ridge regression bias term (1 point)

Linear regression $f(x) = ax + b$ with L2 regularisation on the slope only:
$$a^*, b^* = \arg\min_{a,b} \sum_{i=1}^4 (y_i - f(x_i))^2 + \lambda a^2$$

Data points: $\{(-2, 1), (-1, 3), (0, 2), (3, 4)\}$.

**What is the optimal value for the bias $b^*$?**

- (A) Depends on the value of $\lambda$.
- (B) 3
- **(C) 2.5** ✓
- (D) None of the above.

**Answer: C — $b^* = 2.5$**

**Reason:** The L2 penalty $\lambda a^2$ only penalises the slope $a$. Taking the partial derivative with respect to $b$ and setting it to zero gives $\partial \mathcal{L}/\partial b = -2\sum_i(y_i - ax_i - b) = 0$, which simplifies to $b = \bar{y} - a\bar{x}$. This is exactly the same condition as in ordinary least squares — the bias is unaffected by L2 regularisation on the slope. Since $\bar{y} = (1+3+2+4)/4 = 2.5$ and $\bar{x} = (-2-1+0+3)/4 = 0$, we get $b^* = 2.5$ regardless of $\lambda$.

## Question 4 — PCA (1 point)

**Which transformation to a data matrix $\mathbf{X}$ will affect the principal components obtained through PCA?**

- (A) Adding a constant value to all elements of $\mathbf{X}$.
- **(B) Multiplying one of the features of $\mathbf{X}$ by a constant.** ✓
- (C) Adding an extra feature to $\mathbf{X}$ that is constant across all data points.
- (D) None of the above.

**Answer: B**

**Reason:**
- (A) PCA is performed on the centred data; adding a global constant shifts the mean but does not change the covariance structure → PCs unchanged.
- (B) Multiplying one feature by a constant rescales its variance, which changes the covariance matrix and therefore the eigenvectors (PCs).
- (C) A constant feature has zero variance; it contributes nothing to the covariance matrix and is ignored by PCA.

## Question 5 — Ridge regularisation loss (1 point)

Loss function (note: $\|\cdot\|_2$ norm, not squared):
$$\mathcal{L}_\lambda(\boldsymbol{\beta}) = \frac{1}{N}\sum_{i=1}^N (y_i - \mathbf{x}_i^\top \boldsymbol{\beta})^2 + \lambda\|\boldsymbol{\beta}\|_2$$

Let $C_\lambda = \min_{\boldsymbol{\beta}} \mathcal{L}_\lambda(\boldsymbol{\beta})$.

**Which statement is true?**

- (A) $C_\lambda$ is a non-increasing function of $\lambda$.
- (B) For $\lambda = 0$, the loss is non-convex and might have several minimisers.
- (C) $C_\lambda$ is a non-decreasing function of $\lambda$.
- **(D) None of the above statements are true.** ✓

**Answer: D**

**Reason:**
- (A) False: $C_\lambda$ is generally non-decreasing in $\lambda$ because larger $\lambda$ restricts the feasible region more. But the penalty here is the $\ell_2$ *norm* (not squared), so (C) is also wrong in general.
- (B) False: $\mathcal{L}_0$ is the standard mean squared error, which is convex and has a unique minimiser (or infinitely many if $X$ is rank-deficient — but "non-convex" is incorrect).
- (C) False in general: because the penalty is $\|\boldsymbol{\beta}\|_2$ (not $\|\boldsymbol{\beta}\|_2^2$), the analysis is more subtle and (C) does not hold unconditionally.
- Therefore, none of (A)–(C) is unconditionally true.

## Question 6 — Logistic regression gradient (1 point)

Logistic regression loss for binary classification ($y_i \in \{0,1\}$):
$$\mathcal{L}(\boldsymbol{\beta}) = \frac{1}{N}\sum_{i=1}^N \left(\log(1+e^{\mathbf{x}_i^\top\boldsymbol{\beta}}) - y_i\mathbf{x}_i^\top\boldsymbol{\beta}\right)$$

**Which expression is the gradient $\nabla\mathcal{L}(\boldsymbol{\beta})$?**

- (A) $\frac{1}{N}\sum_i \left(\mathbf{x}_i \frac{e^{\mathbf{x}_i^\top\boldsymbol{\beta}}}{1+e^{\mathbf{x}_i^\top\boldsymbol{\beta}}} - y_i\mathbf{x}_i^\top\boldsymbol{\beta}\right)$
- (B) $\frac{1}{N}\sum_i \mathbf{x}_i\left(y_i - \frac{e^{\mathbf{x}_i^\top\boldsymbol{\beta}}}{1+e^{\mathbf{x}_i^\top\boldsymbol{\beta}}}\right)$
- **(C) $\frac{1}{N}\sum_i \mathbf{x}_i\left(\frac{1}{1+e^{-\mathbf{x}_i^\top\boldsymbol{\beta}}} - y_i\right)$** ✓
- (D) None of the above.

**Answer: C**

**Derivation:** Let $\sigma(z) = 1/(1+e^{-z})$ be the sigmoid. Then:
$$\frac{\partial}{\partial\boldsymbol{\beta}} \log(1+e^{\mathbf{x}_i^\top\boldsymbol{\beta}}) = \frac{e^{\mathbf{x}_i^\top\boldsymbol{\beta}}}{1+e^{\mathbf{x}_i^\top\boldsymbol{\beta}}}\mathbf{x}_i = \sigma(\mathbf{x}_i^\top\boldsymbol{\beta})\mathbf{x}_i$$
So $\nabla\mathcal{L} = \frac{1}{N}\sum_i \mathbf{x}_i(\sigma(\mathbf{x}_i^\top\boldsymbol{\beta}) - y_i)$, which matches (C).

## Question 7 — Irreducible error (1 point)

In linear regression, $p(y \mid \mathbf{x}) \sim \mathcal{N}(\boldsymbol{\beta}^\top\mathbf{x}+\beta_0, \sigma^2)$.

**The irreducible error in this model is:**

- **(A) $\sigma^2$** ✓
- (B) $\mathbb{E}[y\mid\mathbf{x}]$
- (C) $\mathbb{E}[(y - \mathbb{E}[y\mid\mathbf{x}])^2\mid\mathbf{x}]$
- (D) None of the above.

**Answer: A — $\sigma^2$**

**Reason:** The irreducible error (also called noise variance) is the variance of $y$ that cannot be explained by any model of $\mathbf{x}$, i.e. $\text{Var}(\varepsilon) = \sigma^2$. Note that (C) is equal to $\sigma^2$ as well (it is the variance of $y$ around its conditional mean), but (A) is the cleaner direct answer. Note that (B) is the conditional mean, not an error.

## Question 8 — Decision boundary (1 point)

Which classifier could have generated a highly irregular, non-linear decision boundary (as shown in the figure in the exam)?

- (A) Logistic regression

# Part 3: Principal Component Analysis on the `cars04` Dataset (6 points)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Load cars04 dataset — select columns 7:18 (0-indexed), matching R's [, 8:18]
cars_full = pd.read_csv('../cars-fixed04.csv')
cars04 = cars_full.iloc[:, 7:18].copy()

print('Columns selected:', cars04.columns.tolist())
print('Shape:', cars04.shape)
cars04.head()
```

```python
# Standardise (scale=TRUE equivalent) and fit PCA with all 11 components
scaler = StandardScaler()
X_scaled = scaler.fit_transform(cars04)

pca = PCA(n_components=11)
scores = pca.fit_transform(X_scaled)  # equivalent to cars04.pca$x

print('PCA fitted on', X_scaled.shape[0], 'observations and', X_scaled.shape[1], 'features.')
```

```python
# Explained variance table — equivalent to summary(prcomp(cars04, scale=TRUE))
eigenvalues = pca.explained_variance_           # sdev^2 in R
sdev = np.sqrt(eigenvalues)                     # sdev in R
prop_var = pca.explained_variance_ratio_        # Proportion of Variance
cum_var = np.cumsum(prop_var)                   # Cumulative Proportion

pc_labels = [f'PC{i+1}' for i in range(11)]
summary_df = pd.DataFrame({
    'Standard deviation': sdev,
    'Proportion of Variance': prop_var,
    'Cumulative Proportion': cum_var
}, index=pc_labels)

print('Importance of components:')
summary_df.round(4)
```

```python
# Loadings (rotation) for the first 2 PCs — equivalent to cars04.pca$rotation[, 1:2]
# pca.components_ has shape (n_components, n_features); transpose to get (n_features, n_components)
loadings = pca.components_.T  # shape: (11, 11)

loadings_df = pd.DataFrame(
    loadings[:, :2],
    index=cars04.columns,
    columns=['PC1', 'PC2']
)
print('Loadings (rotation) for PC1 and PC2:')
loadings_df.round(4)
```

```python
# ── Correlation circle (variable space) ──────────────────────────────────────
# Scale each loading vector by sqrt(eigenvalue) — equivalent to P[,i] <- P[,i] * sqrt(lambda[i])
P = loadings.copy()  # (11, 11)
for i in range(11):
    P[:, i] = P[:, i] * np.sqrt(eigenvalues[i])

fig, ax = plt.subplots(figsize=(6, 6))

# Unit circle
theta = np.linspace(0, 2 * np.pi, 300)
ax.plot(np.cos(theta), np.sin(theta), 'k-', linewidth=0.8)

ax.axhline(0, color='grey', linewidth=0.6)
ax.axvline(0, color='grey', linewidth=0.6)

# Plot and label each variable
offsets_x = [0.06 * (1 if i % 2 == 0 else -1) for i in range(11)]
offsets_y = [-0.06 * (1 if i % 2 == 0 else -1) for i in range(11)]

for j, var in enumerate(cars04.columns):
    ax.scatter(P[j, 0], P[j, 1], color='steelblue', zorder=3)
    ax.text(P[j, 0] + offsets_x[j], P[j, 1] + offsets_y[j], var, fontsize=8)

ax.set_xlim(-1.15, 1.15)
ax.set_ylim(-1.15, 1.15)
ax.set_xlabel('Axis 1 (PC1)')
ax.set_ylabel('Axis 2 (PC2)')
ax.set_title('Correlation circle — Variable space')
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```

```python
# ── Individual space biplot ───────────────────────────────────────────────────
# Row names in the original data (use 'Model' column if it exists, else integer index)
if 'Model' in cars_full.columns:
    row_names = cars_full['Model'].values
elif cars_full.columns[0] in cars_full.columns:
    # Try first column as car name
    row_names = cars_full.iloc[:, 0].values
else:
    row_names = cars_full.index.astype(str).values

highlight = ['Audi RS 6', 'Ford Expedition 4.6 XLT', 'Nissan Sentra 1.8']
all_names = list(row_names)
other_names = [n for n in all_names if n not in highlight]

rng = np.random.default_rng(42)
sample_idx = rng.choice(len(other_names), size=min(25, len(other_names)), replace=False)
sampled_names = [other_names[i] for i in sample_idx]

fig, ax = plt.subplots(figsize=(8, 6))
ax.axhline(0, color='grey', linewidth=0.6)
ax.axvline(0, color='grey', linewidth=0.6)

# Plot sampled background cars
for name in sampled_names:
    idx = all_names.index(name)
    ax.text(scores[idx, 0], scores[idx, 1], name, fontsize=6, color='grey',
            ha='center', va='center')

# Highlight the three specific cars
for name in highlight:
    if name in all_names:
        idx = all_names.index(name)
        ax.text(scores[idx, 0], scores[idx, 1], name, fontsize=10, color='black',
                fontweight='bold', ha='center', va='center')

ax.set_xlim(-8, 8)
ax.set_ylim(-4, 4)
ax.set_xlabel('PC1')
ax.set_ylabel('PC2')
ax.set_title('Individual space — PCA scores')
plt.tight_layout()
plt.show()
```

# Part 4: Spectral Community Detection (6 points)

Questions based on Newman, M.E.J. (2006). *Modularity and community structure in networks.*

---

### (a) Two examples of community detection applications (1 point)

1. **World Wide Web**: Community detection can identify groups of web pages on related topics (e.g., all pages discussing a particular subject cluster together), enabling better search engines and content recommendation.

2. **Social networks**: Community detection can identify social units or communities within platforms like Facebook or LinkedIn — groups of people who interact more frequently with each other than with the rest of the network, corresponding to friend circles, professional groups, or interest communities.

---

### (b) Definition of modularity (1.5 points)

The **modularity** $Q$ of a network partition is, up to a multiplicative constant, the number of edges falling *within* groups minus the **expected** number of within-group edges in a null model (a random network with the same degree sequence):

$$Q = \frac{1}{2m} \sum_{ij} \left(A_{ij} - \frac{k_i k_j}{2m}\right) \delta(c_i, c_j)$$

where $A_{ij}$ is the adjacency matrix, $k_i$ is the degree of vertex $i$, $m$ is the total number of edges, $c_i$ is the community label of vertex $i$, and $\delta$ is the Kronecker delta.

Modularity can be positive or negative. **Positive values** indicate that more edges fall within communities than expected by chance, which is the signature of genuine community structure. Searching for partitions that maximise $Q$ is therefore a principled strategy for finding community structure.

---

### (c) No positive eigenvalues → no community structure (1 point)

When the **modularity matrix** $B$ (where $B_{ij} = A_{ij} - k_i k_j / 2m$) has **no positive eigenvalues**, it means that no bisection of the network can yield a positive modularity gain. In other words, the graph **cannot be satisfactorily split into two communities** and should be left as a single community (no further subdivision is warranted).

---

### (d) Magnitude of leading eigenvector coordinates (1.5 points)

The **leading eigenvector** of the modularity matrix determines the assignment of vertices to the two communities (via the sign of each coordinate). The **magnitude** of each coordinate indicates how strongly that vertex drives the split — vertices with large-magnitude coordinates contribute most to the modularity and are the most confidently assigned to their respective community. Vertices with coordinates close to zero lie near the boundary and their community assignment is uncertain.

---

### (e) Detecting more than two communities — can it find three? (1 point)

Newman proposes a **recursive bisection** procedure:
1. Apply the spectral bisection to the full network to split it into two groups.
2. For each resulting subgroup, apply the spectral bisection again using a submatrix of the modularity matrix appropriate for that subgroup.
3. If the modularity matrix of a subgroup has no positive eigenvalues, that subgroup is not split further (natural stopping criterion).
4. Repeat until no subgroup can be profitably split.

**Yes, it can detect three communities.** The algorithm first splits the network into two; then one of the two sub-networks might itself be further splittable (positive eigenvalues in its sub-modularity matrix), yielding a total of three communities, while the other sub-network is left intact (no positive eigenvalues).
