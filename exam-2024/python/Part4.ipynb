This part continues with the **Karate Club** network dataset and focuses on **spectral community detection** using the modularity matrix.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
```

```python
# Load the Karate Club data
A = np.loadtxt('../datasets/karate_adjacency.csv', delimiter=',')
F = np.loadtxt('../datasets/karate_factions.csv', delimiter=',')

print('Adjacency matrix shape:', A.shape)
print('Number of edges:', int(A.sum() / 2))
```

```python
# ── Modularity matrix ─────────────────────────────────────────────────────────
# D: degree vector (row sums of A)
# m: total number of edges
# B: modularity matrix  B_ij = A_ij - (k_i * k_j) / (2m)
D = A.sum(axis=1)          # degree of each node
m = A.sum() / 2            # total number of edges
B = A - np.outer(D, D) / (2 * m)

print('Modularity matrix B shape:', B.shape)
print('B is symmetric:', np.allclose(B, B.T))
```

```python
# ── Eigendecomposition of the modularity matrix ───────────────────────────────
eigenvalues, eigenvectors = np.linalg.eigh(B)

# Sort by descending eigenvalue
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print('Top 5 eigenvalues:')
print(eigenvalues[:5].round(4))

# Leading eigenvector determines the community split
v1 = eigenvectors[:, 0]
print('\nLeading eigenvector (first 10 entries):')
print(v1[:10].round(4))
```

```python
# ── Community assignment from leading eigenvector ─────────────────────────────
# Assign community based on sign of leading eigenvector
predicted_community = np.where(v1 >= 0, 1, 2)

print('Predicted community assignments (1 or 2):')
print(predicted_community)

# Compare to true faction labels
true_factions = F.astype(int)
accuracy = np.mean(predicted_community == true_factions)
# Note: labels might be flipped — check both
accuracy_flipped = np.mean((3 - predicted_community) == true_factions)
print(f'\nAccuracy vs true factions: {max(accuracy, accuracy_flipped):.4f}')
```
