For each class $y \in \{0, 1\}$, the data is generated as follows:
1. Fix class priors: $p$ for class 1, $1-p$ for class 0.
2. For each data point $i$, sample $y_i \sim \mathcal{B}(p)$: $y_i = 1$ with probability $p$, $y_i = 0$ with probability $1 - p$.
3. Depending on $y_i \in \{0, 1\}$, sample $x_i \in \mathbb{R}^2$:
   - $y_i = 0 \Rightarrow x_i \sim \mathcal{N}(\mu_0, \Sigma_0)$
   - $y_i = 1 \Rightarrow x_i \sim \mathcal{N}(\mu_1, \Sigma_1)$

where $\mu_0 = [0, 0]^\top$, $\mu_1 = [\varepsilon, 0]^\top$, $\Sigma_0 = 0.5\, I_2$, $\Sigma_1 = 0.4\, I_2$.
