# Part 3: Open Questions

This part uses the **Karate Club** network dataset.
- `karate_adjacency.csv`: adjacency matrix of the network
- `karate_factions.csv`: faction labels (1 or 2) for each node

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import networkx as nx
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Load the Karate Club data
A = np.loadtxt('../datasets/karate_adjacency.csv', delimiter=',')
F = np.loadtxt('../datasets/karate_factions.csv', delimiter=',')

print('Adjacency matrix shape:', A.shape)
print('Faction labels shape:', F.shape)
print('Unique factions:', np.unique(F))
```

```python
# Build NetworkX graph from adjacency matrix
G = nx.from_numpy_array(A)

print(f'Number of nodes: {G.number_of_nodes()}')
print(f'Number of edges: {G.number_of_edges()}')
```

```python
# Plot the network coloured by faction
faction_labels = F.astype(int)  # 1 or 2
color_map = ['steelblue' if f == 1 else 'tomato' for f in faction_labels]

fig, ax = plt.subplots(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx(
    G, pos=pos, ax=ax,
    node_color=color_map,
    node_size=300,
    font_size=8,
    edge_color='grey',
    width=0.7
)
ax.set_title('Karate Club network coloured by faction\n(blue = faction 1, red = faction 2)')
ax.axis('off')
plt.tight_layout()
plt.show()
```
