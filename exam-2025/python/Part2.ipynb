## (a) Dataset simulation and scatter plot (2.0 points)

Simulate a dataset of size $N = 1000$. For a given $i$, what is the distribution of $(X_{1,i}, X_{2,i})$? Plot the cloud of points of the simulated values $(X_{1,i}, X_{2,i})_{i=1,\dots,N}$. What is its shape? Can you write a closed-form expression for it?

Since the samples are coming from a 2D normal distribution, their scatter plot has an elliptical shape. The closed form expression for their joint PDF is:

$[ X1, X2 ]^T$ ~ N($\mu, \Sigma$) with $\mu$ = $[ 0, 0 ]^T$ and $\Sigma= [[1,4],[4,17]]$

To obtain the expression for the elliptical shape of the scatter points, one can check for the level curves of the pdf, ie the locus of points for which N($\mu,\Sigma$) = k for different values of k $\in R$. We get then different sets of $x^T=(x_1,x_2)^T\in R^2$ for which:
$$\frac{1}{Z}exp(-\frac{1}{2}x^T\Sigma^{-1}x)=k <=> x^T\Sigma^{-1}x=l <=> 17x_1^2-8x_1x_2+x^2=n$$
for real constants k, l, n
