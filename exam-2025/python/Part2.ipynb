The data-generating model is:
$$
\begin{array}{rcl}
X_{1,i} &=& \varepsilon_{1,i} \\
X_{2,i} &=& 4X_{1,i} + \varepsilon_{2,i} \\
Y_{i} &=& X_{2,i} + X_{1,i} - 5 + \varepsilon_{3,i}
\end{array}

where $\varepsilon_{ij} \overset{\text{iid}}{\sim} \mathcal{N}(0,1)$.

Yᵢ = X₂,ᵢ + X₁,ᵢ − 5 + ε₃,ᵢ

where i ∈ {1, …, N} and the εᵢⱼ are independent N(0, 1) random variables. For a given i, what is the distribution of (X₁,ᵢ, X₂,ᵢ)? Plot the cloud of points of the simulated values. What is its shape? Can you write a closed-form expression for it? **(2 points)**

## (a) Dataset simulation and scatter plot (2.0 points)

Simulate a dataset of size $N = 1000$. For a given $i$, what is the distribution of $(X_{1,i}, X_{2,i})$? Plot the cloud of points of the simulated values $(X_{1,i}, X_{2,i})_{i=1,\dots,N}$. What is its shape? Can you write a closed-form expression for it?

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

def generate_dataset(N, seed=0):
    np.random.seed(seed)
    X1 = np.random.randn(N)
    X2 = 4*X1 + np.random.randn(N)
    Y = X2 + X1 - 5 + np.random.randn(N)
    df = pd.DataFrame()
    df['X1'] = X1
    df['X2'] = X2
    df['Y'] = Y
    return df

df = generate_dataset(N=1000, seed=0)

fig, ax = plt.subplots(figsize=(4, 4))
ax.scatter(df['X1'], df['X2'], c='C0', alpha=0.25)
ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.set_title('Scatter plot of (X1, X2) for N=1000')
plt.show()
```

**Answer (a):**

Since $X_1 \sim \mathcal{N}(0,1)$ and $X_2 = 4X_1 + \varepsilon_2$ with $\varepsilon_2 \sim \mathcal{N}(0,1)$ independent of $X_1$, the pair $(X_1, X_2)$ follows a bivariate Normal distribution:
$$
\begin{bmatrix} X_1 \\ X_2 \end{bmatrix} \sim \mathcal{N}(\boldsymbol{\mu}, \boldsymbol{\Sigma})
\quad\text{with}\quad
\boldsymbol{\mu} = \begin{bmatrix} 0 \\ 0 \end{bmatrix},
\quad
\boldsymbol{\Sigma} = \begin{bmatrix} 1 & 4 \\ 4 & 17 \end{bmatrix}
$$

The scatter plot has an **elliptical shape**. The level curves of the pdf (i.e. the locus of points where $\mathcal{N}(\boldsymbol{\mu}, \boldsymbol{\Sigma}) = k$ for constant $k$) satisfy:
$$
\mathbf{x}^\top \boldsymbol{\Sigma}^{-1} \mathbf{x} = \ell
\iff
17x_1^2 - 8x_1 x_2 + x_2^2 = \eta
$$
for real constants $\ell, \eta$, which defines a family of ellipses.

## (b) Simple regression models A and B (2.0 points)

Consider two regression models:
- **Model A**: $Y_i = \alpha_1 X_{1,i} + \alpha_0 + \tilde{\varepsilon}_{A,i}$
- **Model B**: $Y_i = \beta_2 X_{2,i} + \beta_0 + \tilde{\varepsilon}_{B,i}$

What should the values of $\hat{\alpha}_0, \hat{\alpha}_1, \hat{\sigma}_A^2, \hat{\beta}_0, \hat{\beta}_2, \hat{\sigma}_B^2$ be when $N \to \infty$? Fit both models with $N = 1000$ and check the estimates. Then repeat with `seed=3`, $N = 10$.

```python
import statsmodels.api as sm

df = generate_dataset(N=1000, seed=0)
df['intercept'] = 1

model = sm.OLS(df['Y'], df[['intercept', 'X1']])
results_A = model.fit()
print('Model A (N=1000)')
print(results_A.params)
print('MSE residual:', results_A.mse_resid)

model = sm.OLS(df['Y'], df[['intercept', 'X2']])
results_B = model.fit()
print('\nModel B (N=1000)')
print(results_B.params)
print('MSE residual:', results_B.mse_resid)
```

```python
df_small = generate_dataset(N=10, seed=3)
df_small['intercept'] = 1

model = sm.OLS(df_small['Y'], df_small[['intercept', 'X1']])
results_A_small = model.fit()
print('Model A (N=10, seed=3)')
print(results_A_small.params)
print('MSE residual:', results_A_small.mse_resid)

model = sm.OLS(df_small['Y'], df_small[['intercept', 'X2']])
results_B_small = model.fit()
print('\nModel B (N=10, seed=3)')
print(results_B_small.params)
print('MSE residual:', results_B_small.mse_resid)
```

**Answer (b):**

When $N \to \infty$:
- **Model A**: $\alpha_1 = 5,\ \alpha_0 = -5,\ \sigma_A^2 = 2$
- **Model B**: $\beta_2 = \tfrac{5}{4},\ \beta_0 = -5,\ \sigma_B^2 = 1 + \tfrac{1}{16}$

With $N = 1000$ the coefficient estimates are close to the true values, but the residual variance estimates may differ. With $N = 10$ the estimates deteriorate significantly because the sample is too small — the variance of the estimators is much larger.

## (c) Full model with X1 and X2 (1.0 point)

Consider the full model:
$$
Y_i = \gamma_2 X_{2,i} + \gamma_1 X_{1,i} + \gamma_0 + \varepsilon_i
$$

For the simulated data with $n = 10$ (seed=3), estimate $\hat{\gamma}_0, \hat{\gamma}_1, \hat{\gamma}_2, \hat{\sigma}^2$. Compare with (b) and discuss the effects of $X_1$ and $X_2$ on $Y$ and their correlation.

```python
df_small = generate_dataset(N=10, seed=3)
df_small['intercept'] = 1

model = sm.OLS(df_small['Y'], df_small[['intercept', 'X1', 'X2']])
results_C = model.fit()
print('Model C (N=10, seed=3)')
print(results_C.params)
print('MSE residual:', results_C.mse_resid)
print(results_C.summary())
```

**Answer (c):**

The results with $N = 10$ are quite poor. When we perform statistical inference on the coefficients, we may fail to reject the null hypothesis for $\gamma_1$, giving the false impression that $X_1$ has no influence on $Y$, when by construction it does. This effect is mainly due to:
1. The strong **correlation between $X_1$ and $X_2$** (multicollinearity): the model cannot separate their individual contributions.
2. The very **small sample size** ($N = 10$): estimators have high variance, and confidence intervals are wide.
