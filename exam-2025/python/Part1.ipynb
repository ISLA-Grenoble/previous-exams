## Figure for Question 7

Run the cell below to generate the figure illustrating the initial dataset and four possible preprocessing results (A, B, C, D).

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import StandardScaler

N = 100
x1 = 4 + np.random.randn(N)
x2 = 12 * (np.random.rand(N) - 0.5)
X = np.c_[x1, x2]

fig = plt.figure(figsize=(13.0, 5.9))
ax1 = plt.subplot2grid((2,4),(0,0),rowspan=2,colspan=2)
ax1.scatter(X[:,0],X[:,1],c='C0',s=30)
ax1.set_xlabel('Feature X1'); ax1.set_ylabel('Feature X2')
ax1.set_title('Initial dataset'); ax1.set_aspect('equal')
ax1.set_xticks(np.arange(-6,7,2)); ax1.set_yticks(np.arange(-6,7,2))
ax1.set_xlim(-6,6); ax1.set_ylim(-6,6); ax1.grid(True)

ax2 = plt.subplot2grid((2,4),(0,2)); scaler=StandardScaler(); X_A=scaler.fit_transform(X)
ax2.scatter(X_A[:,0],X_A[:,1],c='C0',s=10); ax2.set_title('Preprocessing A')
ax2.set_aspect('equal'); ax2.set_xlim(-6,6); ax2.set_ylim(-6,6); ax2.grid(True)

ax3 = plt.subplot2grid((2,4),(0,3)); X_B=X-X.mean(axis=0)
ax3.scatter(X_B[:,0],X_B[:,1],c='C0',s=10); ax3.set_title('Preprocessing B')
ax3.set_aspect('equal'); ax3.set_xlim(-6,6); ax3.set_ylim(-6,6); ax3.grid(True)

ax4 = plt.subplot2grid((2,4),(1,2)); scaler=StandardScaler(); X_C=scaler.fit_transform(X); X_C=X_C+scaler.mean_
ax4.scatter(X_C[:,0],X_C[:,1],c='C0',s=10); ax4.set_title('Preprocessing C')
ax4.set_aspect('equal'); ax4.set_xlim(-6,6); ax4.set_ylim(-6,6); ax4.grid(True)

ax5 = plt.subplot2grid((2,4),(1,3)); scaler=StandardScaler(); X_D=scaler.fit_transform(X); X_D=X_D*0.5+1.0
ax5.scatter(X_D[:,0],X_D[:,1],c='C0',s=10); ax5.set_title('Preprocessing D')
ax5.set_aspect('equal'); ax5.set_xlim(-6,6); ax5.set_ylim(-6,6); ax5.grid(True)
plt.tight_layout(); plt.show()
```

**Q7 Answer**: (A) Preprocessing A

`StandardScaler` subtracts the mean and divides by the standard deviation of each feature. This centers the data at zero and normalises the spread to unit variance, which matches Preprocessing A.
